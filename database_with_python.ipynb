## SQL and SQLite

```python
import sqlite3
```

```python
## connent to an SQLite database
connection = sqlite3.connect('example.db')
connection
```

```python
## make cursur object   ## cursor performs sql commands
cursor = connection.cursor()
```

```python
## create a table
# Create a table
cursor.execute('''
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
)
''')

## commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
## Insert some data
cursor.execute('''
Insert Into employees(name, age, department)
values("Alireza",34, "Data Engineering")

''')

cursor.execute('''
Insert Into employees(name, age, department)
values("John",45, "Sales")

''')

cursor.execute('''
Insert Into employees(name, age, department)
values("Sara",22, "Marketing")

''')

connection.commit()
```

```python
## query the data from the table
cursor.execute('Select * from employees')
rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## updata data in the table

cursor.execute('''
UPDATE employees
Set age = 30
WHERE name = "John"
''')

connection.commit()
```

```python
## query the data from the table
cursor.execute('Select * from employees')
rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## delet the data from the table
cursor.execute('''
delete from employees
Where name = 'Sara'
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## working with Sales data

connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

## create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INTeGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
    )
''')

## insert data into the sales table
## list and inner is tuples
sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North')
]

```

```python
cursor.execute("ALTER TABLE sales ADD COLUMN sales INTEGER")
connection.commit()
```

```python
cursor.execute("ALTER TABLE sales ADD COLUMN region TEXT")
connection.commit()
```

```python
cursor.executemany("""
INSERT INTO sales (date, product, sales, region)
VALUES (?, ?, ?, ?)
""", sales_data)

connection.commit()
```
