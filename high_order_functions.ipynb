```python
def cubeofnumber(x):
    return x**3
```

```python
def squarerootofnumber(x):
    return x**0.5
```

```python
def squareofanumber(x):
    return x*x
```

```python
squareofanumber(4)
```

```python
def squareofalist(l1):
    l1squared = []
    for i in l1:
        l1squared.append(squareofanumber(i))


    return l1squared
```

```python
squareofalist([1,2,3,4])
```

```python
def squareofalist(l1, functionwhichyouwanttoapply):
    l1squared = []
    for i in l1:
        l1squared.append(functionwhichyouwanttoapply(i))


    return l1squared
```

```python
squareofalist([1,2,3,4], squarerootofnumber)
```

```python
squareofalist([1,2,3,4], cubeofnumber)
```

##### Write a program to get a list and return each and every element as a positive or absolute value

```python
import math
```

```python
def factorialofanumber(x):
    return math.factorial(x)
```

```python
def operationonlist(l1, func):
    l1modified = []
    for i in l1:
        l1modified.append(func(i))

    return l1modified
```

```python
operationonlist([1,2,3,4], factorialofanumber)
```

```python
def positive(x):
    return x >= 0
```

```python
def filtermylist(l1, func):
    l1modified = []
    for i in l1:
        if(func(i)):
           l1modified.append(i)
    
    return l1modified
```

```python
filtermylist([1,2,3,-6,-4,9], positive)
```

```python
## another example

def shout(text):
    return text.upper()

def whisper(text):
    return text.lower()

def greet(func):
    return func("Alireza")
```
