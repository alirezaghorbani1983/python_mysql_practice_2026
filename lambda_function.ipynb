### Lambda function

```python
def square(x):
    return x*x
```

```python
def applyonlist(l1, func):
    l2 = []
    for i in l1:
        l2.append(func(i))

    return l2
```

```python
applyonlist([1,2,3,4,5], square)
```

## Map, filter and Reduce

```python
def square(x):
    return x*x
```

```python
def applyonlist(func,l1):
    l2 = []
    for i in l1:
        l2.append(func(i))

    return l2
```

```python
map(square, [1,2,3,4])
```

```python
list(map(square, [1,2,3,4]))
```

```python
tuple(list(map(square, [1,2,3,4])))
```

```python
tuple(list(map(lambda x:x*x, [1,2,3,4])))
```

```python
tuple(list(map(lambda x:x*x*x, [1,2,3,4])))
```

```python
tuple(list(map(lambda x:x+5, [1,2,3,4])))
```

### Filter

```python
## filter function

def FilterMyList(func, l1):
    l1Modified = []
    for i in l1:
        if(func(i)):
            l1Modified.append(i)

    return l1Modified
```

```python
def iseven(num):
    return num%2 ==0
```

```python
FilterMyList(iseven, [1,2,3,4])
```

```python
f1 = list(filter(iseven, [1,2,3,4]))
```

```python
list(filter(lambda x: x>10, [1,4,88,7,12]))
```
