### Lambda function

```python
def square(x):
    return x*x
```

```python
def applyonlist(l1, func):
    l2 = []
    for i in l1:
        l2.append(func(i))

    return l2
```

```python
applyonlist([1,2,3,4,5], square)
```

```python
def cube(x):
    return x*x*x
```

```python
def applyonlist(l1, func):
    l2 = []
    for i in l1:
        l2.append(func(i))

    return l2
```

```python
applyonlist([1,2,3,4,5], cube)
```

```python
def squareroot(x):
    return x**(1/2)
```

```python
def applyonlist(l1, func):
    l2 = []
    for i in l1:
        l2.append(func(i))

    return l2
```

```python
applyonlist([1,2,3,4,5], squareroot)
```

```python
## lambda function

def square(x):
    return x*x
```

```python
lambda x:x*x
```

```python
sqlambda = lambda x:x*x
```

```python
sqlambda(8)
```

```python
applyonlist([1,2,3,4,5],  lambda x:x*x)
```

```python
applyonlist([1,2,3,4,5], sqlambda)
```

```python
applyonlist([1,2,3,4,5],  lambda x:x*x*x)
```

```python
applyonlist([1,2,3,4,5],  lambda x:x**(0.5))
```

```python
applyonlist([1,2,3,4,5],  lambda x:  abs(x))
```
