## Dictionaries

```python
## creating dictionaries

empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"alireza", "age":42, "grade": 89}
print(student)
print(type(student))
           
```

```python
## single key is always used
student = {"name":"alireza", "age":42, "name": 89}
print(student)
```

```python
## accessing dictionary elements
student = {"name":"alireza", "age":42, "grade": "A"}
print(student)

```

```python
print(student['grade'])
print(student['age'])
```

```python
## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
```

```python
## modifying dictionary elements

print(student)
```

```python
student["age"] = 33 ## updated value for the key
print(student)
student["address"]= "India"  ## added a new key and value
print(student)
```

```python
## delete the key

del student['grade']
print(student)
```

```python
## dictionary methods

keys = student.keys()  ## get all the keys
print(keys)
values = student.values()  ## get all the values
print(values)

items = student.items()  ## get all key values
print(items)
```

```python
## shallow copy

student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "krish"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy()   ## shallow copy
print(student_copy1)
print(student)
```

```python
student["name"] = "Tom"
print(student_copy1)
print(student)
```

```python
## iterating over dictionaries

## iterating over keys

for key in student.keys():
    print(key)
```

```python
## iterate over values
for value in student.values():
    print(value)
```

```python
## iterate over values
for key,value in student.items():
    print(f"{key}: {value}")
```

```python
## Nested dictionaries

students = {
    "student1" : {"name": "ali", "age": 32},
    "student2" : {"name":"John", "age": 55}

}

print(students)
```

```python
## access nested dictionaries elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

#### Dictionary comprehension

```python
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionary comprehension

evens = {x:x**2 for x in range(10) if x%2 == 0}
print(evens)
```

```python
## practical examples

## use adictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number] = 1

print(frequency)
```

```python
## merge 2 dictionaries into one

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged_dict = {**dict1, **dict2}
print(merged_dict)
```
