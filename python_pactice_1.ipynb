## Dictionaries

```python
## creating dictionaries

empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"alireza", "age":42, "grade": 89}
print(student)
print(type(student))
           
```

```python
## single key is always used
student = {"name":"alireza", "age":42, "name": 89}
print(student)
```

```python
## accessing dictionary elements
student = {"name":"alireza", "age":42, "grade": "A"}
print(student)

```

```python
print(student['grade'])
print(student['age'])
```

```python
## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
```

```python
## modifying dictionary elements

print(student)
```

```python
student["age"] = 33 ## updated value for the key
print(student)
student["address"]= "India"  ## added a new key and value
print(student)
```

```python
## delete the key

del student['grade']
print(student)
```

```python
## dictionary methods

keys = student.keys()  ## get all the keys
print(keys)
values = student.values()  ## get all the values
print(values)

items = student.items()  ## get all key values
print(items)
```

```python
## shallow copy

student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "krish"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy()   ## shallow copy
print(student_copy1)
print(student)
```

```python
student["name"] = "Tom"
print(student_copy1)
print(student)
```
